```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
x=pd.read_csv(r"C:\Users\shivam parihari\Downloads\titanic dataset.csv")
```

```python
x.info()
```

```python
x.isnull().sum()
x.head()
```

```python
d=x.describe(include='all')
d
```

```python
x['Age'].fillna(x['Age'].mean(),inplace=True)
x.isnull().sum()
```

```python
x.drop(columns='Cabin',inplace=True)
```

```python
x['Embarked'].fillna(x['Embarked'].mode()[0], inplace=True)
```

```python
x.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
```

```python
x['Sex']=le.fit_transform(x['Sex'])
x['Name']=le.fit_transform(x['Name'])
x['Ticket']=le.fit_transform(x['Ticket'])
x['Embarked']=le.fit_transform(x['Embarked'])
```

```python
x.head()
```

```python
ip=x.drop(columns='Survived')
```

```python
op=x.Survived
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(ip,op,train_size=0.8)
```

```python
from sklearn.linear_model import LogisticRegression
model=LogisticRegression()
model.fit(xtrain,ytrain)
```

```python
ypred=model.predict(xtest)
ypred
```

```python
from sklearn.metrics import accuracy_score,confusion_matrix,classification_report
```

```python
accuracy_score(ytest,ypred)
```

```python
confusion_matrix(ytest,ypred)
```

```python
print(classification_report(ytest,ypred))
```
