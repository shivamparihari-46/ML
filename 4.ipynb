```python
import numpy as np
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
import warnings
warnings.filterwarnings('ignore')
```

```python
x=pd.read_csv(r"C:\Users\shivam parihari\Downloads\titanic dataset.csv")
x.head()
```

```python
x.isnull().sum()
```

```python
x['Age'].fillna(x['Age'].mean(),inplace=True)
x.isnull().sum()
```

```python
d=x.describe(include='all')
d
```

```python
x['Cabin'].fillna('?',inplace=True)
x['Embarked'].fillna('?',inplace=True)
```

```python
x['Embarked']=x['Embarked'].replace('?',d['Embarked'][2])
x['Cabin']=x['Cabin'].replace('?',d['Cabin'][2])
```

```python
x.isnull().sum()
```

```python
from sklearn.preprocessing import LabelEncoder
le=LabelEncoder()
x['Sex']=le.fit_transform(x['Sex'])
x['Name']=le.fit_transform(x['Name'])
x['Ticket']=le.fit_transform(x['Ticket'])
x['Cabin']=le.fit_transform(x['Cabin'])
x['Embarked']=le.fit_transform(x['Embarked'])
```

```python
ip=x.drop(columns='Survived')
ip
op=x.Survived
```

```python
from sklearn.model_selection import train_test_split
xtrain,xtest,ytrain,ytest=train_test_split(ip,op,train_size=0.85)
```

```python
from sklearn.preprocessing import StandardScaler

standard=StandardScaler()
standard.fit_transform(xtrain)
standard.fit_transform(xtest)
```

```python
from sklearn.svm import SVC
```

```python
model=SVC(kernel='rbf')
```

```python
model.fit(xtrain,ytrain)
```

```python
pred=model.predict(xtest)
```

```python
from sklearn.metrics import accuracy_score
accuracy_score(ytest,pred)
```

```python
from sklearn.model_selection import cross_val_score
```

```python
ip1=standard.fit_transform(ip)
```

```python
scores=cross_val_score(model,ip1,op,cv=5,scoring='accuracy')
print(scores)
# scores is the array of accuracy scores when we got by cross validation
```
